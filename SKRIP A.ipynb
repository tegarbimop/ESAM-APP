```python
import os
import shutil

kaggle_dir = '/content/drive/MyDrive/ESAM APP/dataset_waste_kaggle'
esam_organik_dir = '/content/drive/MyDrive/ESAM APP/dataset_esam/organik'

os.makedirs(esam_organik_dir, exist_ok=True)

moved_count = 0

for root, dirs, files in os.walk(kaggle_dir):
    for file in files:
        if file.lower().endswith(('.jpg', '.jpeg', '.png', '.webp')):
            src_file = os.path.join(root, file)
            # Ambil semua gambar dari folder yang bernama organic/food/bio/dll.
            if any(k in root.lower() for k in ['organ', 'food', 'bio', 'compost', 'veg', 'fruit']):
                dst_file = os.path.join(esam_organik_dir, f"kaggle_org_{moved_count}_{file}")
                shutil.copy(src_file, dst_file)
                moved_count += 1

print(f"BERHASIL! Menambahkan {moved_count} gambar organik ke Drive!")
```
